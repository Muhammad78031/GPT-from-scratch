```python

import os
import copy
import torch
import torch.nn as nn
import torch.nn.functional as F
import tiktoken
from google.colab import drive

drive.mount('/content/drive')

device = "cuda" if torch.cuda.is_available() else "cpu"
checkpoint_path = "/content/drive/MyDrive/gpt_checkpoint.pt"

vocab_size = 50257
block_size = 64
embed_dim = 128
num_heads = 4
num_layers = 4

class Head(nn.Module):
    def __init__(self, head_size, embed_dim, block_size):
        super().__init__()
        self.key = nn.Linear(embed_dim, head_size, bias=False)
        self.query = nn.Linear(embed_dim, head_size, bias=False)
        self.value = nn.Linear(embed_dim, head_size, bias=False)
        self.register_buffer("tril", torch.tril(torch.ones(block_size, block_size)))
        self.dropout = nn.Dropout(0.0)

    def forward(self, x, past_kv=None):
        B, T, C = x.shape
        k = self.key(x)
        q = self.query(x)
        v = self.value(x)

        if past_kv is None:
            past_length = 0
            all_k, all_v = k, v
        else:
            past_k, past_v = past_kv
            past_length = past_k.shape[1]
            all_k = torch.cat((past_k, k), dim=1)
            all_v = torch.cat((past_v, v), dim=1)

        wei = q @ all_k.transpose(-2, -1)
        wei = wei * (k.shape[-1] ** -0.5)
        mask = self.tril[past_length:past_length + T, :past_length + T]
        wei = wei.masked_fill(mask == 0, float("-inf"))
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)

        return wei @ all_v, (all_k, all_v)


class MultiHeadAttention(nn.Module):
    def __init__(self, num_heads, head_size, embed_dim, block_size):
        super().__init__()
        self.heads = nn.ModuleList([
            Head(head_size, embed_dim, block_size)
            for _ in range(num_heads)
        ])
        self.proj = nn.Linear(num_heads * head_size, embed_dim)

    def forward(self, x, past_kvs=None):
        outputs, new_past_kvs = [], []

        for i, head in enumerate(self.heads):
            past_kv = None if past_kvs is None else past_kvs[i]
            head_out, present_kv = head(x, past_kv)
            outputs.append(head_out)
            new_past_kvs.append(present_kv)

        return self.proj(torch.cat(outputs, dim=-1)), new_past_kvs


class FeedForward(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(embed_dim, 4 * embed_dim),
            nn.ReLU(),
            nn.Linear(4 * embed_dim, embed_dim)
        )

    def forward(self, x):
        return self.net(x)


class Block(nn.Module):
    def __init__(self, embed_dim, num_heads, block_size):
        super().__init__()
        head_size = embed_dim // num_heads
        self.sa = MultiHeadAttention(num_heads, head_size, embed_dim, block_size)
        self.ffwd = FeedForward(embed_dim)
        self.ln1 = nn.LayerNorm(embed_dim)
        self.ln2 = nn.LayerNorm(embed_dim)

    def forward(self, x, past_kvs=None):
        attn_out, new_past_kvs = self.sa(self.ln1(x), past_kvs)
        x = x + attn_out
        x = x + self.ffwd(self.ln2(x))
        return x, new_past_kvs


class GPTLanguageModel(nn.Module):
    def __init__(self, vocab_size, embed_dim, block_size, num_heads, num_layers):
        super().__init__()
        self.block_size = block_size
        self.token_embedding_table = nn.Embedding(vocab_size, embed_dim)
        self.position_embedding_table = nn.Embedding(block_size, embed_dim)
        self.blocks = nn.ModuleList([
            Block(embed_dim, num_heads, block_size)
            for _ in range(num_layers)
        ])
        self.ln_f = nn.LayerNorm(embed_dim)
        self.lm_head = nn.Linear(embed_dim, vocab_size, bias=False)
        self.token_embedding_table.weight = self.lm_head.weight

    def forward(self, idx, targets=None, past_kvs_per_block=None):
        B, T = idx.shape
        past_length = (
            0 if past_kvs_per_block is None
            else past_kvs_per_block[0][0][0].shape[1]
        )

        tok_emb = self.token_embedding_table(idx)
        positions = torch.arange(past_length, past_length + T, device=idx.device)
        x = tok_emb + self.position_embedding_table(positions)

        new_past_kvs_per_block = []

        for i, block in enumerate(self.blocks):
            block_past = (
                None if past_kvs_per_block is None
                else past_kvs_per_block[i]
            )
            x, new_block_past = block(x, block_past)
            new_past_kvs_per_block.append(new_block_past)

        logits = self.lm_head(self.ln_f(x))
        loss = None

        if targets is not None:
            loss = F.cross_entropy(
                logits.reshape(B * T, -1),
                targets.reshape(B * T),
                ignore_index=-100
            )

        return logits, loss, new_past_kvs_per_block


assert os.path.exists(checkpoint_path), f"Checkpoint nahi mili: {checkpoint_path}"

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=False
)

policy_model = GPTLanguageModel(
    vocab_size, embed_dim, block_size, num_heads, num_layers
).to(device)

policy_model.load_state_dict(checkpoint["model_state_dict"])

reference_model = copy.deepcopy(policy_model).to(device)

for param in reference_model.parameters():
    param.requires_grad = False

policy_model.train()
reference_model.eval()

optimizer = torch.optim.AdamW(
    policy_model.parameters(),
    lr=1e-5
)

enc = tiktoken.get_encoding("gpt2")

print("Checkpoint loaded successfully")
print("Policy model ready")
print("Reference model frozen")
print("Device:", device)
```

```python
import torch
import torch.nn.functional as F
def sequence_logprob(model,prompt,answer):
  prompt_ids=enc.encode(prompt)
  answer_ids=enc.encode(answer)
  prompt_ids = prompt_ids[:block_size - 1]
  answer_ids = answer_ids[:block_size - len(prompt_ids)]
  all_ids=prompt_ids+answer_ids
  input_ids = torch.tensor(all_ids[:-1], dtype=torch.long).unsqueeze(0).to(device)
  targets = torch.tensor(all_ids[1:], dtype=torch.long).to(device)
  logits, _, _ = model(input_ids)
  log_probs = F.log_softmax(logits[0], dim=-1)
  token_log_probs = log_probs.gather(
        dim=-1,
        index=targets.unsqueeze(-1)
    ).squeeze(-1)
  start = len(prompt_ids) - 1
  end = start + len(answer_ids)
  return token_log_probs[start:end].sum()
  item = preference_data[0]

chosen_logp = sequence_logprob(
    policy_model, item["prompt"], item["chosen"]
)

rejected_logp = sequence_logprob(
    policy_model, item["prompt"], item["rejected"]
)

print("Chosen log probability:", chosen_logp.item())
print("Rejected log probability:", rejected_logp.item())
```

```python
beta = 0.1
epochs = 5

for epoch in range(epochs):
    total_loss = 0

    for item in preference_data:
        optimizer.zero_grad(set_to_none=True)


        policy_chosen = sequence_logprob(
            policy_model, item["prompt"], item["chosen"]
        )
        policy_rejected = sequence_logprob(
            policy_model, item["prompt"], item["rejected"]
        )


        with torch.no_grad():
            ref_chosen = sequence_logprob(
                reference_model, item["prompt"], item["chosen"]
            )
            ref_rejected = sequence_logprob(
                reference_model, item["prompt"], item["rejected"]
            )


        dpo_logits = beta * (
            (policy_chosen - ref_chosen)
            - (policy_rejected - ref_rejected)
        )

        loss = -F.logsigmoid(dpo_logits)


        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            policy_model.parameters(), 1.0
        )
        optimizer.step()

        total_loss += loss.item()

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Loss: {total_loss / len(preference_data):.4f}"
    )
```

```python
def generate_text(model, prompt, max_new_tokens=50):
    model.eval()

    context = torch.tensor(
        [enc.encode(prompt)],
        dtype=torch.long
    ).to(device)

    with torch.no_grad():
        for _ in range(max_new_tokens):

            idx = context[:, -block_size:]

            logits, _, _ = model(idx)


            logits = logits[:, -1, :]


            probs = F.softmax(logits, dim=-1)


            next_token = torch.multinomial(probs, num_samples=1)


            context = torch.cat((context, next_token), dim=1)

    return enc.decode(context[0].tolist())
```

```python

import time
import torch

prompt = "The king said"
max_new_tokens = 30
runs = 3

def benchmark(model, name):

    for _ in range(1):
        generate_text(model, prompt, max_new_tokens=5)

    times = []

    for _ in range(runs):
        start = time.perf_counter()

        generate_text(
            model,
            prompt,
            max_new_tokens=max_new_tokens
        )

        end = time.perf_counter()
        times.append(end - start)

    avg_time = sum(times) / len(times)

    print(f"{name} average time: {avg_time:.3f} seconds")
    return avg_time

fp32_time = benchmark(model_fp32, "FP32")
int8_time = benchmark(model_int8, "INT8")

print("\nSpeed comparison:")

if int8_time < fp32_time:
    print(f"INT8 is {fp32_time / int8_time:.2f}x faster")
else:
    print(f"INT8 is {int8_time / fp32_time:.2f}x as fast as FP32")
```

```python
import time
import torch
import torch.nn.functional as F

policy_model.eval()

prompt = "The King said"
max_new_tokens = 200

prompt_ids = torch.tensor(
    [enc.encode(prompt)],
    dtype=torch.long,
    device=device
)

@torch.no_grad()
def generate_without_cache(model, prompt_ids, max_new_tokens):
    idx = prompt_ids.clone()

    for _ in range(max_new_tokens):
        context = idx[:, -block_size:]
        logits, _, _ = model(context)
        next_token_logits = logits[:, -1, :]
        probs = F.softmax(next_token_logits, dim=-1)
        next_token = torch.multinomial(probs, num_samples=1)
        idx = torch.cat((idx, next_token), dim=1)

    return idx


@torch.no_grad()
def generate_with_cache(model, prompt_ids, max_new_tokens):
    idx = prompt_ids.clone()
    past_kvs = None

    logits, _, past_kvs = model(
        idx, past_kvs_per_block=None
    )

    for _ in range(max_new_tokens):
        next_token_logits = logits[:, -1, :]
        probs = F.softmax(next_token_logits, dim=-1)
        next_token = torch.multinomial(probs, num_samples=1)
        idx = torch.cat((idx, next_token), dim=1)

        past_length = past_kvs[0][0][0].shape[1]

        if past_length >= model.block_size:
            context = idx[:, -model.block_size:]
            logits, _, past_kvs = model(
                context, past_kvs_per_block=None
            )
        else:
            logits, _, past_kvs = model(
                next_token, past_kvs_per_block=past_kvs
            )

    return idx


_ = generate_without_cache(policy_model, prompt_ids, 5)
_ = generate_with_cache(policy_model, prompt_ids, 5)


policy_model.eval()

start = time.time()
out_no_cache = generate_without_cache(
    policy_model, prompt_ids, 200
)
no_cache_time = time.time() - start

print("Without KV Cache:", round(no_cache_time, 2), "seconds")

start = time.time()
out_cache = generate_with_cache(
    policy_model, prompt_ids, 200
)
cache_time = time.time() - start

print("With KV Cache:", round(cache_time, 2), "seconds")

if cache_time > 0:
    print("Speedup:", round(no_cache_time / cache_time, 2), "x")
```

```python

import copy
import torch
import torch.nn.functional as F

beta = 0.1
epochs = 5
lr = 1e-5
save_path = "/content/drive/MyDrive/gpt_dpo_checkpoint.pt"

policy_model.eval()
reference_model.eval()

for p in reference_model.parameters():
    p.requires_grad = False

optimizer = torch.optim.AdamW(policy_model.parameters(), lr=lr)

def get_sequence_logprob(model, prompt, answer):
    device = next(model.parameters()).device
    prompt_ids = enc.encode(prompt)
    answer_ids = enc.encode(answer)

    ids = prompt_ids + answer_ids
    ids = ids[:model.block_size]

    prompt_len = min(len(prompt_ids), len(ids))
    if len(ids) < 2 or prompt_len >= len(ids):
        return None

    x = torch.tensor([ids[:-1]], dtype=torch.long, device=device)
    targets = torch.tensor(ids[1:], dtype=torch.long, device=device)

    logits, _, _ = model(x)
    log_probs = F.log_softmax(logits[0], dim=-1)

    start = max(prompt_len - 1, 0)
    positions = torch.arange(start, len(ids) - 1, device=device)
    target_tokens = targets[start:]

    return log_probs[positions, target_tokens].sum()

@torch.no_grad()
def evaluate_preferences(model):
    model.eval()
    results = []

    for item in preference_data:
        chosen = get_sequence_logprob(model, item["prompt"], item["chosen"])
        rejected = get_sequence_logprob(model, item["prompt"], item["rejected"])

        if chosen is not None and rejected is not None:
            results.append((chosen.item(), rejected.item()))

    if not results:
        return {"count": 0, "chosen_higher": 0, "mean_margin": 0.0}

    margins = [c - r for c, r in results]
    return {
        "count": len(margins),
        "chosen_higher": sum(m > 0 for m in margins),
        "mean_margin": sum(margins) / len(margins)
    }

before = evaluate_preferences(policy_model)
print("Before DPO:", before)

reference_scores = []
for item in preference_data:
    with torch.no_grad():
        chosen = get_sequence_logprob(
            reference_model, item["prompt"], item["chosen"]
        )
        rejected = get_sequence_logprob(
            reference_model, item["prompt"], item["rejected"]
        )
    reference_scores.append((chosen, rejected))

policy_model.train()

for epoch in range(epochs):
    total_loss = 0.0
    valid_examples = 0

    for i, item in enumerate(preference_data):
        ref_chosen, ref_rejected = reference_scores[i]

        if ref_chosen is None or ref_rejected is None:
            continue

        policy_chosen = get_sequence_logprob(
            policy_model, item["prompt"], item["chosen"]
        )
        policy_rejected = get_sequence_logprob(
            policy_model, item["prompt"], item["rejected"]
        )

        if policy_chosen is None or policy_rejected is None:
            continue

        chosen_ratio = policy_chosen - ref_chosen
        rejected_ratio = policy_rejected - ref_rejected

        loss = -F.logsigmoid(
            beta * (chosen_ratio - rejected_ratio)
        )

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(policy_model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()
        valid_examples += 1

    avg_loss = total_loss / max(valid_examples, 1)
    print(f"Epoch {epoch + 1}/{epochs} | DPO loss: {avg_loss:.4f}")

after = evaluate_preferences(policy_model)
print("\nAfter DPO:", after)

torch.save({
    "model_state_dict": policy_model.state_dict(),
    "beta": beta,
    "epochs": epochs,
    "preference_data": preference_data
}, save_path)

print("\nDPO checkpoint saved:", save_path)
```

```python

import torch
import torch.nn.functional as F

policy_model.eval()

@torch.no_grad()
def chatbot(prompt, max_new_tokens=40, temperature=0.8):
    device = next(policy_model.parameters()).device

    input_ids = torch.tensor(
        [enc.encode(prompt)],
        dtype=torch.long,
        device=device
    )

    for _ in range(max_new_tokens):
        context = input_ids[:, -policy_model.block_size:]

        logits, _, _ = policy_model(context)
        next_token_logits = logits[:, -1, :] / temperature
        next_token_logits = logits[:, -1, :] / temperature
        next_token_logits[:, 198] = -float("inf")

        probs = F.softmax(next_token_logits, dim=-1)
        next_token = torch.multinomial(probs, num_samples=1)

        probs = F.softmax(next_token_logits, dim=-1)
        next_token = torch.argmax(probs, dim=-1, keepdim=True)

        input_ids = torch.cat((input_ids, next_token), dim=1)

    generated_ids = input_ids[0].tolist()
    return enc.decode(generated_ids)

test_prompts = [
    "The king said",
    "Once upon a time",
    "The best way to learn is"
]

for prompt in test_prompts:
    print("\nPrompt:", prompt)
    print("Response:", chatbot(prompt))
```
